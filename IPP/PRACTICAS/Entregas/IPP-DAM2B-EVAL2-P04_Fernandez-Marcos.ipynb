```python
# EJERCICIO 5 ##########################################################################################################
import random

posicion_jonathan = 0
posicion_mary = 0
longitud_carrera = 0
mensaje_estado_carrera = """
– El atleta Jonathan Smit se encuentra en la posición: {jonathan} 
– El atleta Mary Thompson se encuentra en la posición: {mary}
"""

while True :
    longitud_carrera = int(input("Introduzca la longitud de la carrera (entre 200 y 800): "))

    if longitud_carrera < 0 :
        print("Longitud de la carrera inválida. Por favor vuelva a introducirla.")
        continue
    if longitud_carrera < 200 :
        print("Longitud de la carrera demasiado corta. Por favor vuelva a introducirla.")
        continue
    if longitud_carrera > 800 :
        print("Longitud de la carrera demasiado larga. Por favor vuelva a introducirla.")
        continue
    
    print(f"La longitud de la carrera es: {longitud_carrera}.")
    break

print(mensaje_estado_carrera.format(jonathan=posicion_jonathan, mary=posicion_mary))
while True :
    posicion_jonathan = posicion_jonathan + random.randint(1, 10)
    posicion_mary = posicion_mary + random.randint(1, 10)

    print(mensaje_estado_carrera.format(jonathan=posicion_jonathan, mary=posicion_mary))

    if posicion_jonathan > longitud_carrera :
        break
    if posicion_mary > longitud_carrera :
        break

    continue

if posicion_jonathan > posicion_mary :
    print("Jonathan Smit ha ganando la carrera.")
elif posicion_mary > posicion_jonathan :
    print("Mary Thompson ha ganando la carrera.")
else :
    print("Mary Thompson y Jonathan Smit han empatado.")
```

```python
# EJERCICIO 6 ##########################################################################################################

# Definiciones #####
def conversor_a_binario(numero: int, primera_vez: bool = True) -> str :

    if numero == 0 :
        if primera_vez == True :
            return "0"
        else :
            return ""
    
    return conversor_a_binario(numero // 2, False) + str(numero % 2)

decimal: int
binario: str
####################

while True :
    decimal = int(input("Introduzca un número entero positivo (0, 1, 2, 3,...): "))

    if decimal < 0 :
        print("El número introducido no es positivo. Por favor, introduzca otro número.")
        continue

    break

binario = conversor_a_binario(decimal)

print(f"El {decimal} en binario sería {binario}.")
```

```python
# EJERCICIO 7 ##########################################################################################################

# Definiciones #####
def factorial(numero: int, cantidad: int = 1) -> int :
    if numero == 0 or numero == 1:
        return cantidad
    return factorial(numero - 1, cantidad * numero)

num: int
fact: int
####################

while True : 
    num = int(input("Introduzca un número entero positivo (0, 1, 2, 3,...): "))
    
    if num < 0 :
        print("El número introducido no es positivo. Por favor, introduzca otro número.")
        continue

    break

fact = factorial(num)

print(f"El factorial de {num} es {fact}.")
```

```python
# EJERCICIO 8 ##########################################################################################################

# Definiciones #####
def fibonacci(profundidad: int, numero_1: int = 0, numero_2: int = 1) :
    if profundidad == 0 :
        return 0
    if profundidad == 1 :
        return numero_2
    
    return fibonacci(profundidad - 1, numero_2, numero_2 + numero_1)

posicion: int
miembro: int
####################

while True : 
    posicion = int(input("Introduzca el miembro de la secuencia de Fibonacci que desee saber (0, 1, 2, 3,...): "))
    
    if posicion < 0 :
        print(f"El {posicion} no es un miembro válido. Por favor, vuelva a escoger.")
        continue

    break

miembro = fibonacci(posicion)

print(f"El miembro de la serie de Fibonacci en la posicion {posicion} es el {miembro}.")
```

```python
# EJERCICIO 9 ##########################################################################################################
import random
import enum

# Definiciones #####
class Mensaje (enum.Enum) :
    INICIAL = "Vamos a jugar a que adivines mi número favorito."

    ENTRADA_DIFICULTAD = "¿Qué dificultad quieres Fácil (F) o Difícil (D)?: "
    ERROR_DIFICULTAD = "No entiendo qué dificultad has escogido."
    MODO_FACIL = "Has escogido el modo fácil."
    MODO_DIFICIL = "Has escogido el modo difícil."

    ENTRADA_INTENTO = "Intenta adivinar mi número favorito (1-10): "
    ERROR_INTENTO = "No has introducido un número entre 1 y 10. Vuelve a intentarlo."
    PISTA_MAYOR_INTENTO = "Mi número es mayor."
    PISTA_MENOR_INTENTO = "Mi número es menor."
    INTENTOS_DIFICL = "Te quedan {intentos} intentos."
    INTENTOS_FACIL = "Has realizado {intentos} intentos."
    VICTORIA = "Enhorabuena, has adivinado mi número."
    DERROTA = "Lo siento, no has adivinado mi número a tiempo."

modo_dificil: bool
victoria_usuario: bool

intentos_maximos: int = 3
numero_de_intentos: int = 0

numero_generado: int
intento_actual: int
####################

numero_generado = random.randint(1, 10)

print(Mensaje.INICIAL.value)

while True :
    entrada = input(Mensaje.ENTRADA_DIFICULTAD.value)

    if entrada.__eq__("F") :
        modo_dificil = False
        break
    if entrada.__eq__("D") :
        modo_dificil = True
        break

    print(Mensaje.ERROR_DIFICULTAD.value)
    continue

if modo_dificil :
    print(Mensaje.MODO_DIFICIL.value)
else :
    print(Mensaje.MODO_FACIL.value)

while True :
    entrada = input(Mensaje.ENTRADA_INTENTO.value)

    if not entrada.isdecimal() :
        print(Mensaje.ERROR_INTENTO.value)
        continue
    entrada = int(entrada)
    if entrada < 1 or entrada > 10 :
        print(Mensaje.ERROR_INTENTO.value)
        continue

    intento_actual = entrada
    intentos_maximos = intentos_maximos - 1
    numero_de_intentos = numero_de_intentos + 1

    if intento_actual == numero_generado :
        victoria_usuario = True
        break
    if modo_dificil and intentos_maximos < 1 :
        victoria_usuario = False
        break

    if numero_generado > intento_actual :
        print(Mensaje.PISTA_MAYOR_INTENTO.value)
    else :
        print(Mensaje.PISTA_MENOR_INTENTO.value)

    if modo_dificil :
        print(Mensaje.INTENTOS_DIFICL.value.format(intentos=intentos_maximos))
    else :
        print(Mensaje.INTENTOS_FACIL.value.format(intentos=numero_de_intentos))

    continue

if victoria_usuario :
    print(Mensaje.VICTORIA.value)
else :
    print(Mensaje.DERROTA.value)
```

```python
# EJERCICIO 10 #########################################################################################################

# Definiciones #####
MENSAJE_INICIAL: str = """
Proceda a introducir las notas del exámen.
Su valor tiene que estar entre 0 y 10.
Introducir una nota negativa termina el proceso y muestra la informacion de las notas del exámen
"""
MENSAJE_NOTAS_ENTRADA: str = "Introduzca la siguiente nota: "
MENSAJE_NOTAS_ERROR: str = "La nota introducida no es válida. Vuelve a introducirla"
MENSAJE_NOTAS_FIN: str = "Procediendo a mostrar la información sobre las notas de la clase."
FORMATO_INFO: str = """
### INFO EXÁMEN ###############
 - Nota alta:  {nota_alta}
 - Nota baja:  {nota_baja}
 - Nota media: {nota_media}
 - Alumnos presentados: {alumnos}
"""

nota: float
alumnos: int = 0
nota_media: float = 0.0
nota_alta: float = 0.0
nota_baja: float = 10.0
####################

print(MENSAJE_INICIAL)
while True :
    nota = float(input(MENSAJE_NOTAS_ENTRADA))

    if nota > 10 :
        print(MENSAJE_NOTAS_ERROR)
        continue
    
    if nota < 0 :
        nota_media = nota_media / alumnos
        print(MENSAJE_NOTAS_FIN)
        break

    alumnos = alumnos + 1
    nota_media = nota_media + nota
    if nota_baja > nota :
        nota_baja = nota
    if nota_alta < nota :
        nota_alta = nota
    continue

info_examen = FORMATO_INFO.format(
    nota_baja=nota_baja,
    nota_alta=nota_alta,
    nota_media=nota_media,
    alumnos=alumnos
    )
print(info_examen)
```

```python
# EJERCICIO 13 #########################################################################################################
import random

# Definiciones #########
MENU_PRINCIPAL: str = """
### JUEGO DEL 21 #########
 - Jugar [Enter]
 - Salir [S]
##########################
"""
MENU_JUGADOR: str = """
### {nombre} ############
Puntuación: {pts}
Objetivo:   {obj}
 - Lanzar dados [Enter]
 - Plantarse [P]
##########################
"""
MENU_TIRADA: str = """
### TIRADA ###############
Dado 1: {d1}
Dado 2: {d2}
Puntuación: {pts}
Objetivo: {obj}
##########################
"""
NOMBRE_JUGADOR_1:str = "JUGADOR 1"
NOMBRE_JUGADOR_2:str = "JUGADOR 2"
PTS_OBJETIVO: int = 21

entrada: str
menu: str
pts_jugador_1: int
pts_jugador_2: int
dado_1: int
dado_2: int
gana_jugador_1: bool
gana_jugador_2: bool
########################

while True:
    menu = MENU_PRINCIPAL
    print(menu)
    entrada = input()

    if entrada.__eq__("S") :
        print("Saliendo...")
        break
    print("Empezando juego...")

    pts_jugador_1 = 0
    pts_jugador_2 = 0
    gana_jugador_1 = False
    gana_jugador_2 = False

    while True :
        menu = MENU_JUGADOR.format(
            nombre=NOMBRE_JUGADOR_1,
            pts=pts_jugador_1,
            obj=PTS_OBJETIVO
        )
        print(menu)
        entrada = input()

        if entrada.__eq__("P") :
            print(f"{NOMBRE_JUGADOR_1} se planta.")
            break
        print(f"{NOMBRE_JUGADOR_1} tira los dados.")
        dado_1 = random.randint(1, 6)
        dado_2 = random.randint(1, 6)
        pts_jugador_1 = pts_jugador_1 + dado_1 + dado_2

        menu = MENU_TIRADA.format(
            d1=dado_1,
            d2=dado_2,
            pts=pts_jugador_1,
            obj=PTS_OBJETIVO
        )
        print(menu)
        input()

        if pts_jugador_1 > PTS_OBJETIVO :
            gana_jugador_2 = True
            break

        continue

    if gana_jugador_2 :
        print(f"Gana {NOMBRE_JUGADOR_2}")
        continue
    
    while True :
        menu = MENU_JUGADOR.format(
            nombre=NOMBRE_JUGADOR_2,
            pts=pts_jugador_2,
            obj=PTS_OBJETIVO
        )
        print(menu)
        entrada = input()

        if entrada.__eq__("P") :
            print(f"{NOMBRE_JUGADOR_2} se planta.")
            break
        print(f"{NOMBRE_JUGADOR_2} tira los dados.")

        dado_1 = random.randint(1, 6)
        dado_2 = random.randint(1, 6)
        pts_jugador_2 = pts_jugador_2 + dado_1 + dado_2

        menu = MENU_TIRADA.format(
            d1=dado_1,
            d2=dado_2,
            pts=pts_jugador_2,
            obj=PTS_OBJETIVO
        )
        print(menu)
        input()

        if pts_jugador_2 > PTS_OBJETIVO :
            gana_jugador_1 = True
            break

        continue

    if gana_jugador_1 :
        print(f"Gana {NOMBRE_JUGADOR_1}")
        
        continue

    if   pts_jugador_1 > pts_jugador_2 :
        print(f"\nGana {NOMBRE_JUGADOR_1}")
    elif pts_jugador_2 > pts_jugador_1 :
        print(f"\nGana {NOMBRE_JUGADOR_2}")
    else :
        print("\nEmpate")
    continue
```

```python
# EJERCICIO 14 #########################################################################################################

# Definiciones #########
DIVISAS: tuple = (
    50000, 20000, 10000,
     5000,  2000,  1000,
      500,   200,   100,
       50,    20,    10,
        5,     2,     1
)

entrada: str
cantidad_euros: float
cantidad_cents: int
lista_desglose: list[int] = []
mensaje_desglose: str = "### DESGLOSE FINAL #########\n"
########################

entrada = input("Intrduzca la cantidad de dinero de la que quiere cambio: ")
cantidad_euros = round(float(entrada), 2)
print(f"Procediendo al desglose de los {cantidad_euros} €\n")
cantidad_cents = int(cantidad_euros * 100)

for divisa in DIVISAS :
    cantidad_divisa = cantidad_cents // divisa
    lista_desglose.append(cantidad_divisa)
    cantidad_cents = cantidad_cents % divisa

    if cantidad_divisa == 0 :
        continue
    mensaje_desglose = mensaje_desglose + f"{cantidad_divisa} divisas de {float(divisa / 100):.2f} €\n"

print(mensaje_desglose)
```
