```python
# EJERCICIO 2 ##########################################################################################################
ica = -1
ica = int(input("Introduzca el Índice de Calidad del Aire (ICA): "))

if ica < 0 :
    print("ERROR: el valor del ICA no puede ser menor que cero.")
elif ica < 51 :
    print("El nivel del ICA es: BUENO.")
elif ica < 101 :
    print("El nivel del ICA es: MODERADO.")
elif ica < 151 :
    print("El nivel del ICA es: NO SALUDABLE PARA GRUPOS SENSIBLES.")
elif ica < 201 :
    print("El nivel del ICA es: INSALUBRE.")
elif ica < 251 :
    print("El nivel del ICA es: MUY INSALUBRE.")
else :
    print("El nivel del ICA es: PELIGROSO.")
```

```python
# EJERCICIO 3 ##########################################################################################################
edad, altura = 0, 0

while True :
    edad = int(input("Introduzca su edad (años): "))
    if edad < 1 or edad > 120 :
        print("La edad introducida es inválida. Por favor, vuelva a introducir su edad y altura.")
        continue

    altura = int(input("Introduzca su altura (cm): "))
    if altura < 1 or altura > 250 :
        print("La altura introducida es inválida. Por favor, vuelva a introducir su edad y altura.")
        continue

    break

print(f"Usted tiene {edad} años y mide {altura} centímetros.")
```

```python
# EJERCICIO 4 ##########################################################################################################
bisiesto = False
anno = 0

while True :
    anno = int(input("Introduzca el año que quiera: "))
    if anno < 0 :
        print("El años introducido no es válido. Por favor vuelva a introducirlo.")
        continue

    break

if anno % 4 == 0 :
    bisiesto = True
    if anno % 100 == 0 :
        bisiesto = False
        if anno % 400 == 0 :
            bisiesto = True

if bisiesto == True :
    print("El año introducido es bisiesto.")
else : 
    print("El año introducido no es bisiesto.")
```

```python
# EJERCICIO 5 ##########################################################################################################
import random

posicion_jonathan = 0
posicion_mary = 0
longitud_carrera = 0
mensaje_estado_carrera = """
– El atleta Jonathan Smit se encuentra en la posición: {jonathan} 
– El atleta Mary Thompson se encuentra en la posición: {mary}
"""

while True :
    longitud_carrera = int(input("Introduzca la longitud de la carrera (entre 200 y 800): "))

    if longitud_carrera < 0 :
        print("Longitud de la carrera inválida. Por favor vuelva a introducirla.")
        continue
    if longitud_carrera < 200 :
        print("Longitud de la carrera demasiado corta. Por favor vuelva a introducirla.")
        continue
    if longitud_carrera > 800 :
        print("Longitud de la carrera demasiado larga. Por favor vuelva a introducirla.")
        continue
    
    print(f"La longitud de la carrera es: {longitud_carrera}.")
    break

print(mensaje_estado_carrera.format(jonathan=posicion_jonathan, mary=posicion_mary))
while True :
    posicion_jonathan = posicion_jonathan + random.randint(1, 10)
    posicion_mary = posicion_mary + random.randint(1, 10)

    print(mensaje_estado_carrera.format(jonathan=posicion_jonathan, mary=posicion_mary))

    if posicion_jonathan > longitud_carrera :
        break
    if posicion_mary > longitud_carrera :
        break

    continue

if posicion_jonathan > posicion_mary :
    print("Jonathan Smit ha ganando la carrera.")
elif posicion_mary > posicion_jonathan :
    print("Mary Thompson ha ganando la carrera.")
else :
    print("Mary Thompson y Jonathan Smit han empatado.")
```
