```python
# EJERCICIO 1 ##########################################################################################################
import math

# Definiciones #########
def inversor_de_texto(n: int, m: int) -> None :
    primos: str = ""
    es_primo: bool

    for a in range(n, m + 1) :
        es_primo = True
        limite = round(math.sqrt(a))
        for b in range(2, limite + 1) :
            if a % b == 0 :
                es_primo = False
                break
        if es_primo :
            primos = primos + str(a) + " "
    print(primos)
    return

n: int
m: int
########################
print("Voy a calcular los primos que hay entre dos números que me digas.")
entrada = input("Dame el primer número: ")
n = int(entrada)
entrada = input("Dame el segundo número: ")
m = int(entrada)

if m < n :
    n, m = m, n

print(f"Los primos entre {n} y {m} son:")
inversor_de_texto(n, m)
```

```python
# EJERCICIO 2 ##########################################################################################################

# Definiciones #########
def busqueda_en_texto(texto: str, busqueda: str) -> int :
    contador = 0

    for letra in texto :
        if letra == busqueda :
            contador = contador + 1
    return contador

texto: str
letra_buscada: str
coincidencias: int
########################

print("Vamos a mirar cuantas veces aparece la letra que quieras en el texto que me pases.")
entrada = input("Dame el texto: ")
texto = entrada
entrada = input("Dame la letra: ")
letra_buscada = entrada

coincidencias = busqueda_en_texto(texto, letra_buscada)
print(f"La letra '{letra_buscada}' aparece {coincidencias} veces en el texto que me has pasado.")
```

```python
# EJERCICIO 4 ##########################################################################################################

# Definiciones #########
matriz: list[list]
entrada: str
columnas: int
filas: int
########################
print("¡Vamos a hacer operaciones con matrices!")
print("Primero ayudame a definir el tamaño de la matriz.")
entrada = input("Dime el número de columnas que tiene: ")
columnas = int(entrada)
entrada = input("Dime el número de filas que tiene: ")
filas = int(entrada)

matriz = [[None for _ in range(columnas)] for _ in range(filas)]

producto: int = 1
print("Ahora ayudame a rellenar la matriz con sus valores.")
for i in range(filas) :
    for j in range(columnas) :
        entrada = input(f"Dame el valor de la posicion ({i},{j}): ")
        valor = int(entrada)

        matriz[i][j] = valor
        producto = producto * valor
        continue
    matriz[i].append(producto)
    producto = 1
    continue

for fila in matriz :
    operandos = fila[:-1] # Todos menos el último elemento
    resultado = fila[-1] # El último elemento
    expresion = " * ".join(str(x) for x in operandos)
    # expresion = " * ".join(map(str, operandos))
    print(f"{expresion} = {resultado}")
```

```python
# EJERCICIO 5 ##########################################################################################################

# Definiciones #########
class Estudiante :
    def __init__(self, nombre: str, apellido: str) :
        self.__nombre = nombre
        self.__apellido = apellido
        self.__notas: dict[str, int | None] = {
            "Prog": None,
            "LLMM": None,
            "BBDD": None,
            "SSII": None,
             "EED": None
        }
    
    @property
    def nombre(self) -> str :
        return self.__nombre
    @property
    def apellido(self) -> str :
        return self.__apellido
    @property
    def nota_prog(self) -> int | None :
        return self.__notas.get("Prog")
    @property
    def nota_llmm(self) -> int | None :
        return self.__notas.get("LLMM")
    @property
    def nota_bbdd(self) -> int | None :
        return self.__notas.get("BBDD")
    @property
    def nota_ssii(self) -> int | None :
        return self.__notas.get("SSII")
    @property
    def nota_eed(self) -> int | None :
        return self.__notas.get("EED")
    
    @nota_prog.setter
    def nota_prog(self, nota: int) -> None :
        self.set_nota("Prog", nota)
    @nota_llmm.setter
    def nota_llmm(self, nota: int) -> None :
        self.set_nota("LLMM", nota)
    @nota_bbdd.setter
    def nota_bbdd(self, nota: int) -> None :
        self.set_nota("BBDD", nota)
    @nota_ssii.setter
    def nota_ssii(self, nota: int) -> None :
        self.set_nota("SSII", nota)
    @nota_eed.setter
    def nota_eed(self, nota: int) -> None :
        self.set_nota("EED", nota)
    
    def set_nota(self, asignatura: str, nota: int) -> None :
        if not nota :
            raise AttributeError("Falta el parámetro 'nota'.")
        if not isinstance(nota, int) :
            raise TypeError("El parámetro 'nota' no es del tipo correcto.")
        if nota < 0 or nota > 10 :
            raise ValueError("El parámetro 'nota' no se encuentra en el rango esperado")
        self.__notas[asignatura] = nota

class Estadisitcas :
    def __init__(self, estudiante: Estudiante) :
        self.__estudiante = estudiante
        self.__media: int = None
        self.__maxima: int = None
        self.__minima: int = None

        
    @property
    def media(self) -> int | None :
        return self.__media
    @property
    def maxima(self) -> int | None :
        return self.__maxima
    @property
    def minima(self) -> int | None :
        return self.__minima
    
    def calcular_media(self) -> None :
        nota_media: int = 0
        nota_media = nota_media + self.__estudiante.nota_prog
        nota_media = nota_media + self.__estudiante.nota_llmm
        nota_media = nota_media + self.__estudiante.nota_bbdd
        nota_media = nota_media + self.__estudiante.nota_ssii
        nota_media = nota_media + self.__estudiante.nota_eed
        nota_media = nota_media / 5
        self.__media = round(nota_media)
    
    def calcular_maximo(self) -> None :
        nota_maxima: int = 0
        if nota_maxima < self.__estudiante.nota_prog :
            nota_maxima = self.__estudiante.nota_prog
        if nota_maxima < self.__estudiante.nota_llmm :
            nota_maxima = self.__estudiante.nota_llmm
        if nota_maxima < self.__estudiante.nota_bbdd :
            nota_maxima = self.__estudiante.nota_bbdd
        if nota_maxima < self.__estudiante.nota_ssii :
            nota_maxima = self.__estudiante.nota_ssii
        if nota_maxima < self.__estudiante.nota_eed :
            nota_maxima = self.__estudiante.nota_eed
        self.__maxima = nota_maxima
    
    def calcular_minimo(self) -> None :
        nota_minima: int = 10
        if nota_minima > self.__estudiante.nota_prog :
            nota_minima = self.__estudiante.nota_prog
        if nota_minima > self.__estudiante.nota_llmm :
            nota_minima = self.__estudiante.nota_llmm
        if nota_minima > self.__estudiante.nota_bbdd :
            nota_minima = self.__estudiante.nota_bbdd
        if nota_minima > self.__estudiante.nota_ssii :
            nota_minima = self.__estudiante.nota_ssii
        if nota_minima > self.__estudiante.nota_eed :
            nota_minima = self.__estudiante.nota_eed
        self.__minima = nota_minima

########################

e1: Estudiante = Estudiante("Marcos", "Fernandez")
e1.nota_prog = 7
e1.nota_eed = 8
e1.nota_bbdd = 3
e1.nota_llmm = 9
e1.nota_ssii = 2
e_e1: Estadisitcas = Estadisitcas(e1)
e_e1.calcular_maximo()
e_e1.calcular_minimo()
e_e1.calcular_media()

print(f"""
ALUMNO:
{e1.nombre} {e1.apellido}
NOTAS:
  Prog: {e1.nota_prog}
  LLMM: {e1.nota_llmm}
  BBDD: {e1.nota_bbdd}
  SSII: {e1.nota_ssii}
  EED:  {e1.nota_eed}
ESTADISTICAS:
  Nota media: {e_e1.media}
  Nota más alta: {e_e1.maxima}
  Nota más baja: {e_e1.minima} 
""")
print()
```

```python
# EJERCICIO 6 ##########################################################################################################
import math
import random

# Definiciones #########
class Figura :
    def __init__(self, color: str) :
        self.__color = color
    
    @property
    def color(self) -> str :
        return self.__color
    
class Circulo (Figura) :
    def __init__(self, color, radio: float) :
        super().__init__(color)
        self.__radio = radio
    
    def calcular_perimetro(self) -> float :
        return 2 * math.pi * self.__radio
    
    def calcular_area(self) -> float :
        return math.pi * self.__radio ** 2

class Rectangulo (Figura) :
    def __init__(self, color: str, ancho: float, largo: float) :
        super().__init__(color)
        self.__lado_sup = largo
        self.__lado_inf = largo
        self.__lado_der = ancho
        self.__lado_izq = ancho
    
    def calcular_perimetro(self) -> float :
        return self.__lado_sup + self.__lado_inf + self.__lado_der + self.__lado_izq
    
    def calcular_area(self) -> float :
        return self.__lado_sup * self.__lado_der

class Cuadrado (Rectangulo) :
    def __init__(self, color: str, lado: float) :
        super().__init__(color, lado, lado)

NUM_FIGURAS: int = 12
LIMITE_INF: int = 4
LIMITE_SUP: int = 12

circulo: Circulo
rectangulo: Rectangulo
cuadrado: Cuadrado
figuras: list[Figura] = []
########################
circulo = Circulo("rojo", 4)
rectangulo = Rectangulo("amarillo", 6, 2)
cuadrado = Cuadrado("verde", 8)

print("CIRCULO")
print(circulo.color)
print(circulo.calcular_perimetro())
print(circulo.calcular_area())
print()
print("RECTANGULO")
print(rectangulo.color)
print(rectangulo.calcular_perimetro())
print(rectangulo.calcular_area())
print()
print("CUADRADO")
print(cuadrado.color)
print(cuadrado.calcular_perimetro())
print(cuadrado.calcular_area())

for i in range(NUM_FIGURAS) :
    figura: Figura
    id_figura = random.randint(0, 2)

    if id_figura == 0 :
        radio = random.random() * (LIMITE_SUP - LIMITE_INF) + LIMITE_INF
        figura = Circulo("rojo", radio)

    if id_figura == 1 :
        ancho = random.random() * (LIMITE_SUP - LIMITE_INF) + LIMITE_INF
        largo = random.random() * (LIMITE_SUP - LIMITE_INF) + LIMITE_SUP
        figura = Rectangulo("azul", ancho, largo)

    if id_figura == 2 :
        lado = random.random() * (LIMITE_SUP - LIMITE_INF) + LIMITE_INF
        figura = Cuadrado("verde", lado)

    figuras.append(figura)


for fig in figuras :
    if isinstance(fig, Circulo) :
        print(f"""
        Círculo {fig.color}:
        Perímetro ... {fig.calcular_perimetro()}
        Área ........ {fig.calcular_area()}
        """)
        continue

    if isinstance(fig, Cuadrado) :
        print(f"""
        Cuadrado {fig.color}:
        Perímetro ... {fig.calcular_perimetro()}
        Área ........ {fig.calcular_area()}
        """)
        continue

    if isinstance(fig, Rectangulo) :
        print(f"""
        Rectángulo {fig.color}:
        Perímetro ... {fig.calcular_perimetro()}
        Área ........ {fig.calcular_area()}
        """)
        continue
```
